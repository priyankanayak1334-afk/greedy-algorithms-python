# Phase: Greedy Algorithms - Assign Cookies Problem

**Objective:** Maximize the number of satisfied children given their hunger/happiness factors and the available cookie sizes.

- **Children ($g$):** Each child $i$ has a greed factor $g[i]$, which is the minimum size of a cookie that the child will be content with.
- **Cookies ($s$):** Each cookie $j$ has a size $s[j]$.
- **Condition:** If $s[j] \ge g[i]$, we can assign the cookie $j$ to the child $i$, and the child will be happy.


## Greedy Strategy

To maximize the number of happy children, we must follow a locally optimal choice at each step:
1. **Sort both arrays:** Sort the children's greed factors (`g`) and the cookie sizes (`s`) in ascending order.
2. **Match the smallest possible cookie:** Iterate through both arrays. For each child, look for the *smallest available cookie* that satisfies their requirement. 
3. **Why this works (Greedy Choice Property):** Satisfying a child with a low greed factor using a massive cookie wastes a resource that could potentially satisfy a greedier child later. By matching the smallest capable cookie to the least demanding child, we preserve larger cookies for children with higher demands.


```python
def findContentChildren(g: list[int], s: list[int]) -> int:
    # 1. Sort both the greed factors and cookie sizes
    g.sort()
    s.sort()
    
    child_ptr = 0
    cookie_ptr = 0
    
    # 2. Iterate through both lists using a two-pointer approach
    while child_ptr < len(g) and cookie_ptr < len(s):
        # If the current cookie can satisfy the current child
        if s[cookie_ptr] >= g[child_ptr]:
            # Move to the next child (this one is satisfied!)
            child_ptr += 1
        
        # Always move to the next cookie (whether used or too small)
        cookie_ptr += 1
        
    # The index of child_ptr represents the total number of happy children
    return child_ptr
```

```python
# Test Case 1: Children want small cookies, but cookies are also small
g1 = [1, 2, 3]
s1 = [1, 1]
print(f"Test 1 Result: {findContentChildren(g1, s1)} (Expected: 1)")

# Test Case 2: Enough large cookies to satisfy multiple children
g2 = [1, 2]
s2 = [1, 2, 3]
print(f"Test 2 Result: {findContentChildren(g2, s2)} (Expected: 2)")

# Test Case 3: Mixed sizes where greedy sorting matters
g3 = [3, 4, 7, 1]
s3 = [2, 1, 8]
print(f"Test 3 Result: {findContentChildren(g3, s3)} (Expected: 2)") # 1 gets 1, 7 gets 8
```
